## Stage 2 – Data Cleaning and Pre-processing

Handle missing values (impute or drop)

Handle duplicates

Treat outliers if required

Check skewness and apply transformations

Convert data types if needed

Feature transformations (date parts, derived fields if required for analysis)

```python
# Clean column names
df.columns = df.columns.str.strip()
```

```python
# Check missing values
df.isnull().sum()
```

```python
# Fill numerical columns with mean
num_cols = ['Rating', 'Delivery Time (min)', 'Customer Service Rating']
for col in num_cols:
    if col in df.columns:
        df[col].fillna(df[col].mean(), inplace=True)
```

```python
# Fill categorical columns with mode
cat_cols = ['Agent Name', 'Review Text', 'Location', 'Order Type',
            'Customer Feedback Type', 'Price Range',
            'Discount Applied', 'Product Availability', 'Order Accuracy']
for col in cat_cols:
    if col in df.columns:
        df[col].fillna(df[col].mode()[0], inplace=True)
```

```python
# Check duplicate rows
df.duplicated().sum()
```

```python
# Remove duplicates
df.drop_duplicates(inplace=True)
```

```python
import numpy as np

# Function to remove outliers using IQR
def remove_outliers(col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    return df[(df[col] >= lower) & (df[col] <= upper)]
```

```python
# Apply for important numerical columns
for col in ['Rating', 'Delivery Time (min)', 'Customer Service Rating']:
    if col in df.columns:
        df = remove_outliers(col)
```

```python
# Check skewness
df[['Rating', 'Delivery Time (min)', 'Customer Service Rating']].skew()
```

```python
# Apply log transformation for skewed column
if 'Delivery Time (min)' in df.columns:
    df['Delivery Time (min)'] = np.log1p(df['Delivery Time (min)'])
```

```python
# Convert categorical columns
for col in cat_cols:
    if col in df.columns:
        df[col] = df[col].astype('category')
```

```python
# Ensure numeric columns are correct
for col in num_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
# Create Rating Category
df['Rating_Category'] = df['Rating'].apply(
    lambda x: 'High' if x >= 4 else 'Medium' if x >= 3 else 'Low'
)
# Create Delivery Speed Category
df['Delivery_Speed'] = df['Delivery Time (min)'].apply(
    lambda x: 'Fast' if x <= 30 else 'Moderate' if x <= 60 else 'Slow'
)
# Convert Discount Applied to binary
df['Discount Applied'] = df['Discount Applied'].map({'Yes': 1, 'No': 0})


# Convert Order Accuracy to binary
df['Order Accuracy'] = df['Order Accuracy'].map({'Correct': 1, 'Incorrect': 0})
```
